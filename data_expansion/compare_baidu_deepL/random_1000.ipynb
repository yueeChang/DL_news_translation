```python
import json
import random


with open('WMT_deen.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

#select 697 random data
sampled_data = random.sample(data, 697)

#write to new file
with open('compare_deen.json', 'w', encoding='utf-8') as f:
    json.dump(sampled_data, f, ensure_ascii=False, indent=4)

print("successfully selected and saved data to compare_deen.json")
```
