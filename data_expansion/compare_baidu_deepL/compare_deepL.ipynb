```python
#entode part
import json

# load the data
with open('compare_enzh.json', 'r', encoding='utf-8') as f:
    compare_data = json.load(f)

# load the translation entode by deepL
with open('deepl_entode.json', 'r', encoding='utf-8') as f:
    deepl_data = json.load(f)

# set a dictionary id | translation
deepl_dict = {entry["id"]: entry for entry in deepl_data}

# select translations with the same id in compare_enzh.json
matched_deepl = []
for entry in compare_data:
    entry_id = entry["id"]
    if entry_id in deepl_dict:
        matched_deepl.append(deepl_dict[entry_id])
    else:
        print(f"warning：ID {entry_id} is not in deepL entode.json")

# save the results
with open('deepl_compare_translated_entode.json', 'w', encoding='utf-8') as f:
    json.dump(matched_deepl, f, ensure_ascii=False, indent=4)

print(f"successfully saved {len(matched_deepl)} translated data")
```
