```python
import csv
import json
input_file = "data_tatoeba.tsv"
output_file = "data_tatoeba.json"
data = []
with open(input_file, "r", encoding="utf-8-sig") as tsv_file:
    reader = csv.reader(tsv_file, delimiter="\t")
    for row in reader:
        data.append({
            "id":int(row[0]),
            "source":row[1],
            "target":row[3]
        })
with open(output_file, "w", encoding="utf-8") as json_file:
    json.dump(data, json_file, ensure_ascii=False, indent=4)
```
